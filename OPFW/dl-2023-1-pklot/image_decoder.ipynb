```python
import numpy as np
import cv2
from PIL import Image
import io
import base64
from IPython.display import display, Image as IPImage
import requests
from flask import Flask, jsonify, request
```

```python
response = requests.get('http://127.0.0.1:5000/inference')
received = response.json()
```

```python
meta = received['results']
for i, data in enumerate(meta):
    print('id: {}, box: {}'.format(data['id'], data['box']))
```

```python
source_str = received['source']
source_bytes = base64.b64decode(source_str)
image = IPImage(data=source_bytes)
display(image)
```

```python
output_str = received['output']
output_bytes = base64.b64decode(output_str)
image = IPImage(data=output_bytes)
display(image)
```
