```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# 데이터 생성 (y = 2x + 3을 학습하는 예제)
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[5.0], [7.0], [9.0], [11.0]])

# 간단한 선형 모델 정의
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # 입력 1, 출력 1

    def forward(self, x):
        return self.linear(x)

model = LinearRegressionModel()

# 손실 함수와 옵티마이저 정의
criterion = nn.MSELoss()  # 평균 제곱 오차
optimizer = optim.SGD(model.parameters(), lr=0.01)  # 확률적 경사 하강법

# 학습 루프
num_epochs = 1000
losses = []  # 손실 값을 저장할 리스트

for epoch in range(num_epochs):
    # 모델 예측
    outputs = model(x)
    
    # 손실 계산
    loss = criterion(outputs, y)
    losses.append(loss.item())  # 손실 값 저장
    
    # 경사 초기화, 역전파, 가중치 업데이트
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    # 100 에포크마다 로그 출력
    if (epoch + 1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

# 결과 확인
print("\n학습 완료 후 예측 결과:")
test_input = torch.tensor([[5.0]])
predicted = model(test_input).item()
print(f"입력: 5.0, 예측 값: {predicted:.4f}")

# 그래프 그리기 (손실 값 변화)
plt.plot(range(num_epochs), losses)
plt.title("Loss over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid()
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.tensorboard import SummaryWriter

# TensorBoard 초기화
writer = SummaryWriter(log_dir="runs/linear_regression")

# 데이터 생성 (y = 2x + 3을 학습하는 예제)
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[5.0], [7.0], [9.0], [11.0]])

# 간단한 선형 모델 정의
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # 입력 1, 출력 1

    def forward(self, x):
        return self.linear(x)

model = LinearRegressionModel()

# 손실 함수와 옵티마이저 정의
criterion = nn.MSELoss()  # 평균 제곱 오차
optimizer = optim.SGD(model.parameters(), lr=0.01)  # 확률적 경사 하강법

# 학습 루프
num_epochs = 1000

for epoch in range(num_epochs):
    # 모델 예측
    outputs = model(x)
    
    # 손실 계산
    loss = criterion(outputs, y)
    
    # 경사 초기화, 역전파, 가중치 업데이트
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    # TensorBoard에 손실 값 기록
    writer.add_scalar("Loss/train", loss.item(), epoch)
    
    # 100 에포크마다 로그 출력
    if (epoch + 1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

# 결과 확인
print("\n학습 완료 후 예측 결과:")
test_input = torch.tensor([[5.0]])
predicted = model(test_input).item()
print(f"입력: 5.0, 예측 값: {predicted:.4f}")

# TensorBoard 종료
writer.close()
```
